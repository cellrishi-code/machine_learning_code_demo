```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import BayesianRidge

X = np.array([
    [1],
    [2],
    [3],
    [4],
    [5],
    [6],
    [7],
    [8]
])

y = np.array([
    35,
    40,
    48,
    55,
    60,
    68,
    75,
    82
])

model = BayesianRidge()

model.fit(X, y)

predictions = model.predict(X)

print("Predictions:")
print(predictions)

print("Coefficient:", model.coef_)
print("Intercept:", model.intercept_)
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression, BayesianRidge

X = np.array([
    [1],
    [2],
    [3],
    [4],
    [5],
    [6],
    [7],
    [8]
])

y = np.array([
    35,
    40,
    48,
    55,
    60,
    68,
    75,
    82
])

linear_model = LinearRegression()
bayesian_model = BayesianRidge()

linear_model.fit(X, y)
bayesian_model.fit(X, y)

linear_prediction = linear_model.predict([[10]])

bayesian_prediction, bayesian_std = bayesian_model.predict(
    [[10]],
    return_std=True
)

print("Linear Regression")
print("Coefficient:", linear_model.coef_[0])
print("Intercept:", linear_model.intercept_)
print("Prediction for x=10:", linear_prediction[0])

print("\nBayesian Regression")
print("Coefficient:", bayesian_model.coef_[0])
print("Intercept:", bayesian_model.intercept_)
print("Prediction for x=10:", bayesian_prediction[0])
print("Prediction uncertainty:", bayesian_std[0])
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(X, y)

prediction = model.predict([[10]])

print(prediction)
```

```python
from sklearn.linear_model import BayesianRidge

model = BayesianRidge()

model.fit(X, y)

prediction, uncertainty = model.predict(
    [[10]],
    return_std=True
)

print(prediction)
print(uncertainty)
```
